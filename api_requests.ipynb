```python
import requests
```

##### requests for reading endpoints

```python
response = requests.get("http://localhost:8000/readings/35")
print(response.json())
print(response.status_code)
```

```python
body = {
  "patient_id": 2,
  "sleep_duration_hours": 8.0,
  "heart_rate": 70,
  "is_deep_sleep": False,
  "raw_device_data": {
    "device": "OuraRing",
    "battery": 82
  },
  "recorded_at": "2026-08-16"
}
response = requests.post("http://localhost:8000/readings", json=body)
print(response.json())
```

```python
response = requests.delete("http://localhost:8000/readings/12")
print(response.json())
```

```python
body = {
    "patient_id": 2,
    "sleep_duration_hours": 8.0,
    "heart_rate": 72,
    "is_deep_sleep": True,
    "raw_device_data": {"device": "OuraRing", "battery": 79},
    "recorded_at": "2026-08-17"
}

response = requests.put("http://localhost:8000/readings/20", json=body)
print(response.status_code)
print(response.json())
```

```python
body = {"heart_rate": 75}  # only updating one field

response = requests.patch("http://localhost:8000/readings/3", json=body)
print(response.status_code)
print(response.json())
```

##### Requests for patients endpoint

```python
response = requests.get("http://localhost:8000/patients/1")
print(response.json())
print(response.status_code)
```

```python
body = {"name": "Jake", "age": 45}
response = requests.post("http://localhost:8000/patients", json=body)
print(response.json())
print(response.status_code)
```

```python
body = {"name": "John", "age": 65}
response = requests.put("http://localhost:8000/patients/6", json=body)
print(response.json())
print(response.status_code)
```

```python
body = {"name": "Jalal"}
response = requests.patch("http://localhost:8000/patients/7", json=body)
print(response.json())
print(response.status_code)
```

```python
response = requests.delete("http://localhost:8000/patients/2")
print(response.json())
print(response.status_code)
```
