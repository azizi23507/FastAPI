```python
import requests
```

```python
response = requests.get("http://localhost:8000/readings/1")
print(response.json())
print(response.status_code)
```

```python
body = {"patient_id": 6, "sleep_duration_hours": 8, "heart_rate": 70, "is_deep_sleep": False }
response = requests.post("http://localhost:8000/readings", json=body)
print(response.json())
```

```python
response = requests.delete("http://localhost:8000/readings/8")
print(response.json())
```

```python
body = {
    "patient_id": 6,
    "sleep_duration_hours": 8.0,
    "heart_rate": 72,
    "is_deep_sleep": True,
    "raw_device_data": {"device": "OuraRing", "battery": 79}
}

response = requests.put("http://localhost:8000/readings/3", json=body)
print(response.status_code)
print(response.json())
```

```python
body = {"heart_rate": 75}  # only updating one field

response = requests.patch("http://localhost:8000/readings/3", json=body)
print(response.status_code)
print(response.json())
```
